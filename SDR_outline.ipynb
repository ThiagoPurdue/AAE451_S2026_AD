# AAE 451: System Definition Review (SDR) Outline
## Project: Next-Generation Carrier-Based Strike Fighter
**Instructor:** Professor Thiago Guimaraes  

---

### 1. Introduction & Executive Summary
* **Title Slide:** Project name, Team number, and all member names.
* **BLUF (Bottom Line Up Front):** A high-level executive summary of the most promising concepts and key mission performance targets met.
* **Design Team Mission Statement:** Define the team's goal, specifically addressing how to break "Augustine's Law" through cost-effective design.
* **Outline Slide:** A roadmap of the 30-minute presentation.

### 2. Requirements, CONOPS & Mission Analysis

* **CONOPS:**
    * Describe the primary market / customer for your future system
    * Develop a CONOPS for a typical flight, including any required ground operations and other required ground equipment.
        * Describe any supporting carrer infrastructure.
        * Describe operation with minimal ground equipment.

* **Mission Profile:**
    * **Air-to-Air:** ___ nm minimum combat radius and Mach ______ dash speed.
    * **Strike:** ______ nm sea-level dash at Mach ________+ with internal/external ordnance.
* **Carrier Suitability Constraints:**
    * Maximum approach speed: ____ knots.
    * Maximum Take-off Gross Weight (TOGW): ______ lb.
    * Physical limits: ___ ft folded wingspan and ______ ft total length.

* **Requirements Matrix:** Show the top design-driven requirements with their rational and target values.

### 3. Technologies and Advanced Concepts
* **a. Promising Advanced Concepts:** Consideration of advanced aircraft concepts with references describing recent work in this area.
* **b. Technology Description (IOC 2035):** Technologies applicable to the strike fighter requirements.
    * **i. TRL Assessment:** Associated Technology Readiness Level (TRL) and justification (must be TRL 6+ per RFP).
    * **ii. Representation:** How your team will represent this technology numerically (e.g., "an all-composite fuselage will reduce empty weight by $X\%$, increase cost by $Y\%$, and reduce skin roughness from $A$ to $B$ [ft]") with references.
    * **iii. Potential benefits and costs.**

### 4. Concept Generation & Selection
* **Generation Process:** Mention the use of brainstorming questions or morphological chart functions used during Week 5.
* **Alternative Concepts:** Showcase 3 most promissing layouts including the main definitions (e.g., wing configurations, tail types, inlet positions).
* **Qualitative Selection:**
    * **Methodology:** Show the methodology used for selection (Pugh’s Method or Weighted Objectives).
    * **Criteria:** Focus on carrier suitability, cost, and mission effectiveness.
* **Selection:** Justification for the 2–3 concepts moving forward for quantitative analysis.

### 5. Sizing & Quantitative Trade Studies
* **Constraint Diagrams:** $T/W$ vs. $W/S$ for all RFP performance points.
* **Quantitative Trades:** Sensitivity on weapon carriage (internal vs. external).

### 6. Selected Propulsion System Definition
* **Engine Selection:** Identification of a selected engine.
* **Specifications:**
    * Number of Engines.
    * Thrust Specification.
    * Weight and dimensions for integration.
    * Specific Fuel Consumption (SFC) values for sizing.
* **Integration:** Impact on Single-Engine Rate of Climb (SEROC).

### 7. Preliminary Performance Calculations
* **Maneuverability & Agility:**
    * **Sustained Turn Rate:** $\omega = $ ______ $~deg/sec$ at ____ ft.
    * **Energy Maneuverability (EM):** Mapping $P_s$ contours to identify combat envelope.
* **Carrier Launch & Recovery:**
    * **Catapult End Speeds ($V_{end}$):** Verification against C13-2 launcher curves.
    * **Wind Over Deck (WOD):** Minimum ship speed required for safe operations.
* **Speed & Altitude Envelopes:**
    * Max Mach ____ dash and Service Ceiling (_________ $+~ft$).
    * Validate your range targets.
    * Calculation of the required internal fuel volume.

### 8. Preliminary Technical Analysis
* **CG & Mass Properties:** Initial weight breakdown and CG range for Air-to-Air vs Strike.
* **Stability Calculations:** Neutral point and static margin estimates.

### 9. Operational and Acquisition Costs
* **a. Detailed Breakdown:** Acquisition costs for the production run of 500 aircraft.
* **b. Operational Analysis:** Analysis of operational costs (fuel, maintenance, crew) over a 25-year service life.
* **c. Cost Methodology:** Students must refer to **"Fundamentals of Aircraft Design" by L.M. Nicolai** for cost calculation methodologies (see Chapter 24 on Life Cycle Costing).
* **d. Student Task:** Compare the **baseline aircraft** with the configurations you are currently developing. Justify how your design maintains or improves performance at a comparable or lower unit acquisition cost.
* **e. Cost Reduction Strategies:** Discussion on breaking the trend of Augustine's Law.


### 10. Conclusion & Path Forward
* **Compliance Matrix.:** Meeting "Target" vs. "Threshold" requirements.
* **Next Steps:** Incorporating feedback, detailing CAD model, optimizing design with carpet plots,  and refining cost estimation, weight, weight and structural design.
