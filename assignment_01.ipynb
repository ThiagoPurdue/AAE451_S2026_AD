You must move beyond basic dimensions and gather the detailed sizing data required to generate an initial OpenVSP or CAD model. Create and populate a **Detailed Geometry Table** (Table 1) with the following:

* **Main Wing:**
    * Reference Area ($S_{ref}$), Aspect Ratio ($AR$), Taper Ratio ($\lambda$).

    * Airfoil types (Root and Tip) or Thickness-to-Chord ratios ($t/c$).
    * Incidence and Dihedral angles.

* **Empennage (Tail Surfaces):**
    * **Horizontal Stabilator:** Area, Span, Sweep, Taper Ratio, and Airfoil/Thickness.
    * **Vertical Stabilizers:** Area (per tail), Height, Sweep, Taper Ratio, and Cant Angle (degrees from vertical).
    * **Tail Arms:** Measure/Estimate the distance between the Main Wing Quarter-Chord ($c/4$) and the Horizontal Tail Quarter-Chord ($L_{ht}$), as well as the Vertical Tail Quarter-Chord ($L_{vt}$). These are critical for longitudinal and directional stability calculations.

* **Fuselage:**
    * Total Length (compare specifically to the RFP 50 ft limit).
    * Max Width and Max Height.
    * **Cross-Section:** Characterize the fuselage cross-section shape (e.g., rectangular, blended) and estimated Fineness Ratio ($L/D_{equiv}$).

* **Landing Gear:**
    * **Type:** Confirm configuration (e.g., Tricycle).
    * **Geometry:** Wheelbase (distance nose-to-main) and Track Width (distance main-to-main).
    * **Tires:** Diameter and Width for both Main and Nose gear (Crucial for packaging).
    * **Struts:** Estimated extended length.
    * **Kinematics:** Verify the "Tip-Back Angle" (angle of rotation around main gear before tail strikes ground).
