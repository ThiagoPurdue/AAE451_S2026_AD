# Report/Presentation Outline: AAE451000-S2026

This document serves as the guide and outline for the Conceptual Design Review (CoDR) oral presentation and Final Design Report.

## 1. Executive Summary & Introduction
* **Title & Team**: Title slide/page with team members.
* **Executive Summary / BLUF**: "Bottom Line Up Front" presentation of the preferred concept, its designation, and a visual (walk-around chart). Highlights of critical outcomes and key parameters.
* **Design Team Mission Statement**: Updated from previous reviews.

## 2. Requirements & Concept of Operations (CONOPS)
* **Missions defined**: Clearly identify the sizing mission and the design missions (Air-to-Air Combat and Strike).
* **CONOPS**: Story from ground preparation (carrier deck operations) to flight finalization, emphasizing the reference mission and system deployment.
* **Compliance Matrix**: Review major aircraft design requirements. Compare the preferred concept against threshold values.

## 3. Concept Generation & Selection
* **Concept Generation / Initial Alternatives**: Overview of brainstorming, morphological charts, and generated concepts.
* **Qualitative Selection**: Detailed Pugh Method / Decision Matrix application demonstrating how qualitative assessments and critical criteria were used to down-select to 2-3 most promising concepts.

## 4. Advanced Aircraft Description & Key Technologies
* **Dimensioned 3-View**: Presentation of the selected final configuration to scale (nose-to-tail, wingspan folded/unfolded, total height, gear footprint).
* **Internal Layout/Inboard Profile**: Representative internal arrangement showing major sub-systems, internal weapons bay, fuel tanks, and primary structures.
* **Key Technologies Implemented**: Highlight technologies, advanced materials, or aerodynamic features used to break "Augustine's Law" and reduce cost or improve performance. Include assumptions and TRL assessment for readiness by 2035 IOC.

## 5. Aircraft Sizing & Optimization
* **Sizing Methodology & Validation**: Description of the sizing methodology (flowchart), drag prediction, and powerplant modeling. Code validation showcasing baseline characterization using the benchmark aircraft (F/A-18E/F).
* **Sizing Studies**: Constraint Diagram and Carpet plots (W/S vs T/W) optimizing the final design point for the preferred concepts. Identifying active performance constraints (e.g., carrier approach speed, sustained turn rate, takeoff distance, SEROC).
* **Quantitative Selection / Down-Select**: Final decision on the optimal design point.

## 6. Mass Properties & CG Envelope
* **Weights Statement**: Aircraft group weights breakdown (structure, propulsion, payload, fuel, Empty Weight, Zero Fuel Weight, ...).
* **Center of Gravity**: Detailed CG build-up, absolute CG envelopes, and CG travel diagrams/unloading profiles (Fig.15.1 Raymer) tracking stability throughout the mission.

## 7. Aerodynamics
* **Planform & Airfoil Selection**: Rationales for wing geometry (sweep, aspect ratio) and airfoil choice. Discussion of trade-offs between subsonic efficiency  and supersonic dash requirements.
* **Subsonic Aerodynamics**:
  * **Span Load & High-Lift Devices**: Subsonic lift distribution during loiter/turn constraints. Sizing of flaps, slats, or other high-lift devices to achieve critical carrier approach speeds.
  * **Subsonic Drag Build-up**: Quantified aerodynamic characteristics (parasite $C_{D,0}$, induced drag factor $K$) across the subsonic envelope.
* **Supersonic Aerodynamics**:
  * **Wave Drag Analysis**: Discuss the methodology utilized and its impact on supersonic drag.
  * **Aerodynamic Center Shift**: Transitioning from subsonic to supersonic regimes, and its effect on longitudinal stability and trim drag.
* **Drag Polars & Performance Summary**: Clean, takeoff, and landing configurations. Plots of $L/D$ vs. Mach bridging both the subsonic and supersonic regimes, clearly contrasting flight efficiency.

## 8. Propulsion System
* **Engine Selection & Integration**: Selection of an existing production engine. Engine modeling approach, thrust lapse, installed performance, and SFC.
* **Inlet Design**: Engine integration and intake/duct geometry (Detailed CAD representation).

## 9. Performance & Flight Envelope
* **Mission Segment Analysis**: Performance estimations for Takeoff/Landing, Climb (including SEROC), Cruise, Loiter, and supersonic Dash.
* **Maneuvering Performance**: Doghouse plots, maximum instantaneous turn rate, sustained turn rate ($>8$ deg/s at 20k ft), and Specific Excess Power (SEP).
* **Flight Envelope & Payload-Range**: Altitude vs. Mach limits and Payload-Range diagrams for both Air-to-Air and Strike combat missions.
* **Carrier Suitability**: Launch considerations (Minimum end airspeed, catapult limits, zero WOD) and Recovery (Arrestment engaging speed, max 10 ft sink rate). Demonstrating deck handling footprint constraints.

## 10. Stability & Control
* **Empennage / Tail Sizing**: Sizing of horizontal/vertical stabilizers and Scissor Plot mapping CG limits.
* **Control Surfaces**: Geometry and sizing of primary flight control surfaces (ailerons, rudder, elevators).
* **Static & Dynamic Stability**: Longitudinal static margin, dynamic stability parameters (time responses, natural frequencies), and validation of stability for all flight phases.
* **Trim Analysis**: Trimmed drag polar definition.

## 11. Structures & Loads
* **V-n Diagram (Loads)**: Dynamic V-n diagram defining structural load limits and design speeds.
* **Structural Layout & Materials**: Primary load paths, wing-box shear/bending moment distribution, and fuselage loads (balanced assembly case). Rationale for material selection in a corrosive maritime environment (25-year service life).
* **Landing Gear**: Specific placement and oleo dimensioning tailored for punishing carrier impact loads, tip-over, and tip-back angles.

## 12. Systems, Avionics, & Survivability
* **Avionics & Sensors**: Integration of the mandatory 2,500 lb sensors/avionics suite required by the RFP (Radar, EW, visual systems).
* **Ordnance / Weapons System**: Demonstration of internal/external carriage geometry ensuring clearance for AIM-120C, AIM-9X, and MK-83 JDAM combinations.
* **Survivability**: Discussion about Radar Cross-Section (RCS) reduction efforts, low observability considerations, and countermeasures (flares/chaffs).

## 13. Cost & Program Lifecycle
* **Aircraft Cost**: Modified DAPCA IV cost analysis evaluating RDTE, manufacturing, and flyaway unit cost for a production run of 500 aircraft, verifying affordability against the baseline F/A-18E/F. Direct operational costs.
* **Design & Development Risk**: Focus on operational/program risks and mitigation to meet the 2035 Initial Operational Capability (IOC).
* **Maintenance & Repair**: Strategy addressing maintenance man-hours (MMH/FH), access panels, and minimizing bespoke support equipment.

## 14. Conclusion & Recommendations
* **Summary**: Final justification of the design's cost-effectiveness and performance, summarizing key successes against RFP demands.
**Design Variants & Future Use-Cases**: Discussion of additional roles, configurations, or potential variants (e.g., electronic attack or non-carrier exports).
* **Future Work**: Actionable recommendations for the next phase of design maturation.
