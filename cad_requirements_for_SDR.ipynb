### 1. Overview

### 3. Required Tasks

#### Task A: Stakeholder Analysis
Before you open your CAD software, ask yourself: *Who is sitting on my SDR Review Board?*
* **The Professional Context:** In a real company, if you don't provide the board with the specific data required to approve a "Go/No-Go" decision, your project will face delays.
* **Your Goal:** Clearly identify what technical evidence they need to feel confident in your design.

#### Task B: Requirement Traceability (Verification & Validation)
Go back to **Assignment 1**. Which of those requirements can we prove with geometry?
* **The Professional Context:** This is the "Verification" part of V&V. If you claim the aircraft has a certain payload volume, the CAD model is your proof.
* **Your Goal:** Create a matrix that links every geometric requirement to a specific feature in your model.

#### Task C: Configuration Assembly
Now, build the high-fidelity assembly. This is more than a 3D sketch; it's a digital prototype.
* **The Professional Context:** Engineers use these models to "freeze" a design.
* **Your Goal:** Produce a model that is clear enough to be used as your primary visual evidence during the SDR.

#### Task D: Project Management
This is where many teams fail. The dimensions in your CAD model must match your analysis code *exactly*.
* **The Professional Context:** This is **Configuration Control**. If the Aerodynamics lead changes the wing chord in their code, but the structures lead doesn't update the model, your team is providing false data. The CAD model should be the source of truth.
* **Your Goal:** Demonstrate a process that ensures your 3D geometry and your analytical variables are always in sync.

**Example Implementation:** Use a shared configuration file (like JSON) to act as the single source of truth for all sub-teams.

```python
import json

# This represents your 'Master Design Variable' file
master_config = {
    "wing_span_m": 12.5,
    "root_chord_m": 1.8,
    "fuselage_length_m": 9.2,
    "engine_diameter_mm": 450,
    "payload_volume_m3": 0.45
}

def verify_cad_sync(cad_measurement, variable_name):
    """Function to simulate a check between CAD and Analysis variables."""
    expected_value = master_config.get(variable_name)
    if cad_measurement == expected_value:
        print(f"[PASS] {variable_name}: CAD matches Analysis ({cad_measurement})")
    else:
        print(f"[FAIL] {variable_name}: DISCREPANCY DETECTED!")
        print(f"       CAD: {cad_measurement}, Analysis: {expected_value}")

# Simulate a check of the wing span
current_cad_span = 12.5 # Replace with actual CAD export data
verify_cad_sync(current_cad_span, "wing_span_m")
```

### 4. General Guidance:
In my experience, the biggest risk to a design team is **isolation**. I often see teams where the "CAD models" work ine corner, while the analysis teams work somewhere else. This *always* results in a frustrating SDR where, for example, you realize the propulsion unit doesn't actually fit inside the wing.

* **Your Strategy:** Utilize the CAD model as a collaborative platform where your ideas come together. Detecting an **interface interference**—when two parts attempt to occupy the same space—is actually a positive, as it indicates you've identified the issue early, in the lab, rather than during the presentation.
